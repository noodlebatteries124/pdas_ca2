```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("(1) graduate employment survey modified.csv")
df1 = pd.read_csv("(2) graduate employment survey modified .csv")

# MERGE ON KEY
df_merged = df.merge(df1, on="Key")
# REMOVE UNNAMED COLUMNS
df_clean = df_merged.loc[:,
    ~df_merged.columns.str.contains('Unnamed', case=False, na=False)
]
df_cleaned = df_clean.loc[:, ~df_clean.T.duplicated()]
print(df_cleaned.info())
# SAVE TO CSV
df_cleaned.to_csv('merged.csv', index=False)

# COPY THE SAVED CSV
df_copy = df_cleaned.copy()
df_copy = df_copy.dropna()
print(df_copy.info())
```
